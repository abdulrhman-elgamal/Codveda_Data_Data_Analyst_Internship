# Level 3 - Task 2: Interactive Dashboard Data Preparation & Merging
**Dataset:** Telecom Customer Churn Dataset (`churn-bigml-80.csv` & `churn-bigml-20.csv`)  
**Objective:** Merge training and testing data splits into a single master dataset, perform preliminary structural validation, and export a clean unified CSV file ready for visual dashboard development in Power BI or Tableau.

```python
# Import pandas for data manipulation and merging
import pandas as pd

# Load the training and testing chunks
df_train = pd.read_csv('churn-bigml-80.csv')
df_test = pd.read_csv('churn-bigml-20.csv')

# Concatenate both dataframes into one master dataset
df_master = pd.concat([df_train, df_test], ignore_index=True)

# Display master dataset basic information
print("--- Master Churn Dataset Summary ---")
print(f"Total rows: {df_master.shape[0]}")
print(f"Total columns: {df_master.shape[1]}")
print("\nMissing values per column:")
print(df_master.isnull().sum().sum())

# Export the master file ready for Power BI / Tableau
df_master.to_csv('telecom_churn_master.csv', index=False, encoding = 'utf-8')
print("\nMaster dataset successfully exported as 'telecom_churn_master.csv'!")

# Preview the first few rows
display(df_master.head())
```
