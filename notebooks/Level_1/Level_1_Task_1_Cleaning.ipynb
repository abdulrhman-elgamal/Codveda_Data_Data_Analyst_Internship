# Level 1 - Task 1: Data Cleaning and Preprocessing
**Dataset:** House Prediction Dataset (`4) house Prediction Data Set.csv`)  
**Objective:** Load the raw dataset, assign meaningful column names, handle missing values, check and remove duplicates, verify data types, and export the cleaned data for the next exploration task.

```python
# Importing Needed Librareies
import pandas as pd

# Define meaningful column names based on the Boston Housing dataset structure
house_columns = [
    'CRIM',
    'ZN',
    'INDUS',
    'CHAS',
    'NOX',
    'RM',
    'AGE',
    'DIS',
    'RAD',
    'TAX',
    'PTRATIO',
    'B',
    'LSTATE',
    'PRICE'
]

# Load the dataset using regular expression separator for white spaces and apply column names
df = pd.read_csv('4) house Prediction Data Set.csv', sep = r'\s+', header = None, names = house_columns)

# Display the first 5 rows to verify the successful loading of data
print('--- First 5 Rows of Dataset ---')
print(df.head())
```

## Step 1: Handling Missing Values
Checking for any null or missing values across all columns in the dataset and dropping them to ensure data quality.

```python
# Check the count of missing values per column
print('--- Missing Values Count Per Column ---')
print(df.isnull().sum())

# Drop rows containing missing values (if any)
df.dropna(inplace = True)
print('\nMissing values have been handled successfully.')
```

## Step 2: Removing Duplicate Rows
Checking for any duplicate rows in the dataset to maintain data integrity and avoid bias.

```python
# Check for Duplicated rows
duplicates_count = df.duplicated().sum()
print(f'Number of duplicate rows found: {duplicates_count}')

# Remove duplicate rows if they exist
df.drop_duplicates(inplace = True)
print('Duplicates removed successfully (if any).')
```

## Step 3: Checking Data Types and Formats
Verifying that all columns have the correct data types (numeric: float/int) after the cleaning process.

```python
# Display dataset structure, non-null counts, and data types
df.info()
```

## Step 4: Exporting the Cleaned Dataset
Saving the processed and cleaned DataFrame into an Excel file to use it seamlessly in Task 2 (EDA and Visualization).

```python
# Export the cleaned dataframe to an Excel file (excluding the dataframe index)
df.to_excel('cleaned_house_data.xlsx', index = False)
print('Dataset successfully cleaned, verified and saved as "cleaned_house_data.xlsx"!')
```
